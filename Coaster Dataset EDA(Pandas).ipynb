# Feature Understanding

Plotting Feature Distribution

```python
df["Year_Introduced"].value_counts() # How manytimes each value of this column came.
```

```python
df["Year_Introduced"].value_counts()\
.head(10)\
.plot(kind="bar")
```

```python
# above view is vertical
# let's see this horizontal
df["Year_Introduced"].value_counts()\
.head(10)\
.plot(kind="barh") # kind="barh"
```

```python
# start doing depth analysis
ax = df["Year_Introduced"].value_counts()\
.head(10)\
.plot(kind="bar",title="Top 10 Years Coasters Introduced")

ax.set_xlabel("Year Introduced")
ax.set_ylabel("Count")
```

```python
# see the distribution of "Speed_mph" column
df["Speed_mph"].plot(kind="hist",bins=20)
```

```python
# More customize
ax=df["Speed_mph"].plot(kind="hist",bins=20,title="Coaster Speed (mph)")
ax.set_xlabel("Speed (mph)")
```

```python
df.head(2)
```

```python
# Q. what are the locations where fastest roller coaster (minimum of 10)?
df["Location"].value_counts()
```

```python
# other locations are not real locations
ax=df.query('Location != "Other"').groupby("Location")["Speed_mph"].agg(["mean","count"]).query('count>=10').sort_values("mean")["mean"]\
.plot(kind="barh",figsize=(12,5),title="Average coaster speed by location")

ax.set_xlabel("Average Coaster Speed")
plt.show()
```
